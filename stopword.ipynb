```python
# =====================================
# main.py
# =====================================

from stopwords import preprocess_text

# =====================================
# TELUGU EXAMPLE
# =====================================

text1 = "ఇది ఒక మంచి ప్రాజెక్ట్ మరియు ఉపయోగకరం"

output1 = preprocess_text(text1, "telugu")

print("TELUGU OUTPUT\n")

print("TOKENS:")
print(output1["tokens"])

print("\nAFTER STOPWORD REMOVAL:")
print(output1["after_stopword_removal"])

print("\nAFTER STEMMING:")
print(output1["after_stemming"])


# =====================================
# MALAYALAM EXAMPLE
# =====================================

text2 = "ഇത് ഒരു നല്ല പ്രോജക്ട് ആണ് കൂടാതെ വളരെ ഉപകാരപ്രദമാണ്"

output2 = preprocess_text(text2, "malayalam")

print("\n\nMALAYALAM OUTPUT\n")

print("TOKENS:")
print(output2["tokens"])

print("\nAFTER STOPWORD REMOVAL:")
print(output2["after_stopword_removal"])

print("\nAFTER STEMMING:")
print(output2["after_stemming"])


# =====================================
# TAMIL EXAMPLE
# =====================================

text3 = "இது ஒரு நல்ல திட்டம் மற்றும் மிகவும் பயனுள்ளது"

output3 = preprocess_text(text3, "tamil")

print("\n\nTAMIL OUTPUT\n")

print("TOKENS:")
print(output3["tokens"])

print("\nAFTER STOPWORD REMOVAL:")
print(output3["after_stopword_removal"])

print("\nAFTER STEMMING:")
print(output3["after_stemming"])

# =====================================
# KANNADA EXAMPLE
# =====================================

text4 = "ಇದು ಒಂದು ಉತ್ತಮ ಪ್ರಾಜೆಕ್ಟ್ ಮತ್ತು ತುಂಬಾ ಉಪಯುಕ್ತವಾಗಿದೆ"

output4 = preprocess_text(text4, "kannada")

print("\n\nKANNADA OUTPUT\n")

print("TOKENS:")
print(output4["tokens"])

print("\nAFTER STOPWORD REMOVAL:")
print(output4["after_stopword_removal"])

print("\nAFTER STEMMING:")
print(output4["after_stemming"])
```
